# Content-Based Filtering

Using `ContentBasedRecommender` (TF-IDF on genres) to recommend movies similar to a given title.

```python
import sys
import os
import joblib
import pandas as pd

sys.path.append('..')
from src.utils.data_loader import load_movielens
from src.content_based.tfidf_recommender import ContentBasedRecommender
```

## 1. Load Data

```python
ratings_df, movies_df = load_movielens(
    ratings_path='../data/raw/u.data',
    movies_path='../data/raw/u.item'
)

print('Ratings shape:', ratings_df.shape)
print('Movies shape:', movies_df.shape)
movies_df.head()
```

## 2. Fit the Content-Based Model

```python
content_model = ContentBasedRecommender()
content_model.fit(movies_df)

print('TF-IDF matrix shape:', content_model.tfidf_matrix.shape)
print('Cosine similarity matrix shape:', content_model.cosine_sim.shape)
```

## 3. Test Recommendations for 5 Different Movies

```python
# Movie 1 — Animation/Children's
print('=== Toy Story (1995) ===')
display(content_model.recommend('Toy Story (1995)', n=10))
```

```python
# Movie 2 — Action/Adventure
print('=== GoldenEye (1995) ===')
display(content_model.recommend('GoldenEye (1995)', n=10))
```

```python
# Movie 3 — Drama
print('=== Fargo (1996) ===')
display(content_model.recommend('Fargo (1996)', n=10))
```

```python
# Movie 4 — Horror/Thriller
print('=== Scream (1996) ===')
display(content_model.recommend('Scream (1996)', n=10))
```

```python
# Movie 5 — Romance/Comedy
print('=== Jerry Maguire (1996) ===')
display(content_model.recommend('Jerry Maguire (1996)', n=10))
```

## 4. Analysis — Do Recommendations Make Sense?

- **Toy Story** → Returns other Animation/Children's/Comedy movies ✅
- **GoldenEye** → Returns other Action/Adventure/Thriller movies ✅
- **Fargo** → Returns Crime/Drama/Thriller movies ✅
- **Scream** → Returns other Horror/Thriller movies ✅
- **Jerry Maguire** → Returns Romance/Comedy/Drama movies ✅

The TF-IDF model correctly groups movies by genre. Limitation: it cannot account for user preferences or movie quality — purely genre-driven.

## 5. Save the Content Model

```python
os.makedirs('../models', exist_ok=True)
joblib.dump(content_model, '../models/content_model.pkl')
print('content_model.pkl saved to models/')
```
